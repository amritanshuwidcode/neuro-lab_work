```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam


# ============================================================
# 1. LOAD DATASET
# ============================================================

data = pd.read_csv("StudentsPerformance.csv")

print("Dataset:")
print(data.head())

print("\nDataset Shape:")
print(data.shape)


# ============================================================
# 2. CREATE BINARY CLASS
# ============================================================

# Math score is used for classification
#
# Math Score >= 40  --> Pass = 1
# Math Score < 40   --> Fail = 0

data["result"] = (data["math score"] >= 40).astype(int)


# ============================================================
# 3. INPUT AND OUTPUT
# ============================================================

# Remove math score because it is used to create the target.
# result is the output/class.

X = data.drop(["math score", "result"], axis=1)

y = data["result"]


# ============================================================
# 4. CONVERT CATEGORICAL DATA INTO NUMERICAL DATA
# ============================================================

X = pd.get_dummies(X, drop_first=True)


# ============================================================
# 5. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# 6. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)


# ============================================================
# 7. CREATE NEURAL NETWORK
# ============================================================

model = Sequential()


# ------------------------------------------------------------
# Hidden Layer 1
# 4 Perceptrons + ReLU
# ------------------------------------------------------------

model.add(
    Dense(
        4,
        activation="relu",
        input_shape=(X_train.shape[1],)
    )
)


# ------------------------------------------------------------
# Hidden Layer 2
# 4 Perceptrons + ReLU
# ------------------------------------------------------------

model.add(
    Dense(
        4,
        activation="relu"
    )
)


# ------------------------------------------------------------
# Hidden Layer 3
# 4 Perceptrons + ReLU
# ------------------------------------------------------------

model.add(
    Dense(
        4,
        activation="relu"
    )
)


# ------------------------------------------------------------
# Single Output Layer
# 1 Neuron
# Sigmoid for Binary Classification
# ------------------------------------------------------------

model.add(
    Dense(
        1,
        activation="sigmoid"
    )
)


# ============================================================
# 8. COMPILE MODEL
# ============================================================

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 9. DISPLAY MODEL ARCHITECTURE
# ============================================================

model.summary()


# ============================================================
# 10. TRAIN THE MODEL
# ============================================================

history = model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)


# ============================================================
# 11. CALCULATE TEST LOSS AND ACCURACY
# ============================================================

loss, accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("\n====================================")
print("TEST RESULTS")
print("====================================")

print("Binary Cross-Entropy Loss =", loss)

print("Accuracy =", accuracy)

print("Accuracy (%) =", accuracy * 100)


# ============================================================
# 12. PREDICTION
# ============================================================

y_probability = model.predict(X_test)

# Probability >= 0.5 --> Pass (1)
# Probability < 0.5  --> Fail (0)

y_pred = (y_probability >= 0.5).astype(int)


# ============================================================
# 13. ACCURACY USING SKLEARN
# ============================================================

final_accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nAccuracy using sklearn =",
      final_accuracy * 100, "%")


# ============================================================
# 14. CONFUSION MATRIX
# ============================================================

print("\nConfusion Matrix:")

cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)


# ============================================================
# 15. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Fail", "Pass"]
    )
)
```
