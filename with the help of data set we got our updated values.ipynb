```python
import numpy as np
import pandas as pd

# Load Dataset
df = pd.read_csv("Student_Marks.csv")
# Data
number_courses = df["number_courses"].values.astype(float)
time_study = df["time_study"].values.astype(float)
actual = df["Marks"].values.astype(float)

# Initial parameters
w1 = 0.01
w2 = 0.10
b = 0.0

learning_rate = 0.0001

rows = []

for i in range(len(df)):

    # Prediction
    prediction = w1 * number_courses[i] + w2 * time_study[i] + b

    # Error
    error = prediction - actual[i]

    # Loss
    loss = error ** 2

    # Gradients
    dw1 = 2 * error * number_courses[i]
    dw2 = 2 * error * time_study[i]
    db = 2 * error

    # Update
    w1 = w1 - learning_rate * dw1
    w2 = w2 - learning_rate * dw2
    b = b - learning_rate * db

    rows.append([
        number_courses[i],
        time_study[i],
        actual[i],
        prediction,
        loss,
        w1,
        w2,
        b
    ])

# Final table
result = pd.DataFrame(rows, columns=[
    "Number of Courses",
    "Time Study",
    "Actual Marks",
    "Predict",
    "Loss",
    "Updated w1",
    "Updated w2",
    "Updated b"
])

print(result)
```
