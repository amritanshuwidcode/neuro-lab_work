```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/MOCK_DATA (1).csv")
print("Dataset:")
print(df.head())

# Input features
X1 = df["iq"].values
X2 = df["CGPA"].values

# Actual output
Y = df["LPA"].values

# ==============================
# 2. Initialize weights and bias
# ==============================

w1 = 0.0
w2 = 0.0
b = 0.0

# Learning rate
lr = 1e-7 # Reduced learning rate to prevent divergence

# Total loss
total_loss = 0

# ==============================
# 3. Process Dataset Row by Row
# ==============================

for i in range(len(df)):

    # Input values
    x1 = X1[i]
    x2 = X2[i]

    # Actual value
    y = Y[i]

    # ------------------------------
    # Prediction
    # ------------------------------

    y_pred = w1 * x1 + w2 * x2 + b

    # ------------------------------
    # Error
    # ------------------------------

    error = y_pred - y

    # ------------------------------
    # Loss for current data point
    # MSE loss = 1/2 (prediction - actual)^2
    # ------------------------------

    loss = 0.5 * (error ** 2)

    # Add loss to total loss
    total_loss += loss

    # ------------------------------
    # Gradient Calculation
    # ------------------------------

    dw1 = error * x1
    dw2 = error * x2
    db = error

    # ------------------------------
    # Weight Update
    # ------------------------------

    w1 = w1 - lr * dw1
    w2 = w2 - lr * dw2
    b = b - lr * db

    # ------------------------------
    # Display each row
    # ------------------------------

    print(
        f"Row {i+1}: "
        f"Actual = {y:.3f}, "
        f"Predicted = {y_pred:.3f}, "
        f"Loss = {loss:.3f}, "
        f"w1 = {w1:.5f}, "
        f"w2 = {w2:.5f}, "
        f"b = {b:.5f}"
    )

# ==============================
# 4. Final Cost Function
# ==============================

cost = total_loss / len(df)

print("\n===================================")
print("FINAL RESULTS")
print("===================================")

print("Final w1 =", w1)
print("Final w2 =", w2)
print("Final bias =", b)

print("Total Loss =", total_loss)
print("Cost Function (MSE) =", cost)
```

```python
#experiment-2

import numpy as np
import pandas as pd


iq = np.array([120, 140, 130, 140], dtype=float)
cgpa = np.array([8, 9, 7, 6], dtype=float)
actual = np.array([7, 9, 5, 6], dtype=float)


w1 = 0.01
w2 = 0.10
b = 0.0

learning_rate = 0.0001

rows = []

for i in range(len(iq)):


    prediction = w1 * iq[i] + w2 * cgpa[i] + b


    error = prediction - actual[i]


    loss = error ** 2


    dw1 = 2 * error * iq[i]
    dw2 = 2 * error * cgpa[i]
    db = 2 * error


    w1 = w1 - learning_rate * dw1
    w2 = w2 - learning_rate * dw2
    b = b - learning_rate * db

    rows.append([
        iq[i],
        cgpa[i],
        actual[i],
        prediction,
        loss,
        w1,
        w2,
        b
    ])

df = pd.DataFrame(rows, columns=[
    "IQ",
    "CGPA",
    "Actual LPA",
    "Predict",
    "Loss",
    "Updated w1",
    "Updated w2",
    "Updated b"
])

print(df)
```
